```python
# ============================================================
# STEP 9 — CAREER EXPLANATION & ROADMAP ENGINE
# ============================================================

import pandas as pd
import os

print("=" * 80)
print("CAREER EXPLANATION & ROADMAP ENGINE")
print("=" * 80)

# ------------------------------------------------------------
# 1. FILE PATHS
# ------------------------------------------------------------

BASE_PATH = r"C:\Users\shiva\Desktop\AI_Career_Recommendation_System"

career_file = os.path.join(
    BASE_PATH,
    "data",
    "career_knowledge_base.csv"
)

jobs_file = os.path.join(
    BASE_PATH,
    "data",
    "processed_jobs.csv"
)

recommendation_file = os.path.join(
    BASE_PATH,
    "data",
    "final_recommendation_test.csv"
)

# ------------------------------------------------------------
# 2. LOAD DATA
# ------------------------------------------------------------

careers = pd.read_csv(career_file)
jobs = pd.read_csv(jobs_file)
recommendations = pd.read_csv(recommendation_file)

print("\nFiles loaded successfully.")

print("Career Knowledge Base:", careers.shape)
print("Processed Jobs:", jobs.shape)
print("Recommendations:", recommendations.shape)


# ------------------------------------------------------------
# 3. CAREER ROADMAP DATABASE
# ------------------------------------------------------------

roadmaps = {

    "AI / Machine Learning Engineer": [
        "Learn Python programming",
        "Learn Mathematics and Statistics",
        "Learn Machine Learning algorithms",
        "Learn Deep Learning",
        "Learn TensorFlow or PyTorch",
        "Build AI/ML projects",
        "Learn model deployment and APIs",
        "Complete internship or real-world project",
        "Apply for AI / Machine Learning Engineer roles"
    ],

    "Data Scientist": [
        "Learn Python and SQL",
        "Learn Statistics and Probability",
        "Learn Data Analysis and Visualization",
        "Learn Machine Learning",
        "Learn Feature Engineering",
        "Learn Deep Learning basics",
        "Build Data Science projects",
        "Create a strong portfolio",
        "Apply for Data Scientist roles"
    ],

    "Data Analyst": [
        "Learn Excel",
        "Learn SQL",
        "Learn Statistics",
        "Learn Python for Data Analysis",
        "Learn Power BI or Tableau",
        "Practice Data Visualization",
        "Build dashboard projects",
        "Create an analytics portfolio",
        "Apply for Data Analyst roles"
    ],

    "Software Developer": [
        "Learn programming fundamentals",
        "Master Python / Java / C++",
        "Learn Data Structures and Algorithms",
        "Learn Object-Oriented Programming",
        "Learn Git and GitHub",
        "Learn databases and SQL",
        "Build software projects",
        "Practice coding interview questions",
        "Apply for Software Developer roles"
    ],

    "Web Developer": [
        "Learn HTML and CSS",
        "Learn JavaScript",
        "Learn React or another frontend framework",
        "Learn backend development",
        "Learn databases and APIs",
        "Learn Git and GitHub",
        "Build responsive websites",
        "Deploy projects online",
        "Apply for Web Developer roles"
    ],

    "Cloud / DevOps Engineer": [
        "Learn Linux fundamentals",
        "Learn networking basics",
        "Learn Git and GitHub",
        "Learn AWS / Azure / Google Cloud",
        "Learn Docker",
        "Learn Kubernetes basics",
        "Learn CI/CD pipelines",
        "Build cloud deployment projects",
        "Apply for Cloud / DevOps roles"
    ],

    "Cybersecurity Analyst": [
        "Learn computer networking",
        "Learn Linux",
        "Learn cybersecurity fundamentals",
        "Learn common cyber attacks",
        "Learn security tools",
        "Learn vulnerability assessment",
        "Practice using cybersecurity labs",
        "Build security projects",
        "Apply for Cybersecurity Analyst roles"
    ],

    "Financial Analyst": [
        "Learn Excel",
        "Learn Financial Statements",
        "Learn Financial Modeling",
        "Learn Statistics",
        "Learn Risk Analysis",
        "Learn SQL and data analysis",
        "Study financial markets",
        "Build financial analysis projects",
        "Apply for Financial Analyst roles"
    ],

    "Digital Marketing Specialist": [
        "Learn Digital Marketing fundamentals",
        "Learn SEO",
        "Learn Social Media Marketing",
        "Learn Google Ads",
        "Learn Content Writing",
        "Learn Market Research",
        "Learn Analytics",
        "Create marketing campaigns/projects",
        "Apply for Digital Marketing roles"
    ],

    "Business Analyst": [
        "Learn Business Analysis fundamentals",
        "Improve communication skills",
        "Learn Excel",
        "Learn SQL",
        "Learn Data Visualization",
        "Learn requirement analysis",
        "Learn problem-solving techniques",
        "Build business case studies",
        "Apply for Business Analyst roles"
    ],

    "Research Scientist": [
        "Build strong Mathematics and Statistics",
        "Learn Python",
        "Learn Research Methodology",
        "Learn Machine Learning",
        "Study scientific literature",
        "Learn experimental design",
        "Work on research projects",
        "Publish or present research work",
        "Apply for Research Scientist roles"
    ]
}


# ------------------------------------------------------------
# 4. FUNCTION TO FIND CAREER INFORMATION
# ------------------------------------------------------------

def get_career_info(career_name):

    career_row = careers[
        careers["career"].str.strip().str.lower()
        == career_name.strip().lower()
    ]

    if career_row.empty:
        return None

    return career_row.iloc[0]


# ------------------------------------------------------------
# 5. FUNCTION TO GET JOB/SALARY INFORMATION
# ------------------------------------------------------------

def get_salary_info(career_name):

    # Search job titles containing important words
    career_words = career_name.lower().replace("/", " ").split()

    matching_jobs = jobs[
        jobs["job_title"]
        .str.lower()
        .apply(
            lambda title: any(
                word in title
                for word in career_words
                if len(word) > 2
            )
        )
    ]

    if matching_jobs.empty:
        return None

    return {
        "average_salary": matching_jobs["salary"].mean(),
        "minimum_salary": matching_jobs["salary"].min(),
        "maximum_salary": matching_jobs["salary"].max(),
        "job_count": len(matching_jobs)
    }


# ------------------------------------------------------------
# 6. GENERATE CAREER EXPLANATION
# ------------------------------------------------------------

def generate_explanation(row):

    career = row["career"]

    info = get_career_info(career)

    if info is None:
        return {
            "career": career,
            "score": row["final_score"],
            "explanation": "Career information not available."
        }

    matching_skills = str(row.get("matching_skills", ""))

    # Convert matching skills into readable list
    if matching_skills.strip():
        skill_list = [
            skill.strip()
            for skill in matching_skills.split(",")
            if skill.strip()
        ]
    else:
        skill_list = []

    # Career skills
    career_skills = [
        skill.strip()
        for skill in str(info["skills"]).split(",")
        if skill.strip()
    ]

    # Missing skills
    missing_skills = [
        skill
        for skill in career_skills
        if skill.lower() not in [x.lower() for x in skill_list]
    ]

    # Salary
    salary_info = get_salary_info(career)

    # Explanation
    explanation_parts = []

    if row["skill_score"] >= 80:
        explanation_parts.append(
            "Your technical skills strongly match this career."
        )
    elif row["skill_score"] >= 50:
        explanation_parts.append(
            "You have a moderate technical skill match for this career."
        )
    else:
        explanation_parts.append(
            "You should develop more technical skills for this career."
        )

    if row["interest_score"] >= 75:
        explanation_parts.append(
            "Your interests strongly align with this career."
        )
    elif row["interest_score"] >= 40:
        explanation_parts.append(
            "Your interests have a reasonable connection with this career."
        )

    if row["aptitude_score"] >= 80:
        explanation_parts.append(
            "Your aptitude profile is highly suitable for this career."
        )

    if row["education_score"] >= 80:
        explanation_parts.append(
            "Your educational background is suitable for this career."
        )

    if row["industry_score"] >= 80:
        explanation_parts.append(
            "Your preferred industry matches this career."
        )

    return {
        "career": career,
        "score": round(row["final_score"], 2),
        "category": info["category"],
        "explanation": " ".join(explanation_parts),
        "matching_skills": matching_skills
            if matching_skills.strip()
            else "No specific matching skills identified",
        "skills_to_improve": ", ".join(missing_skills)
            if missing_skills
            else "No major technical skill gaps identified",
        "education": info["education"],
        "programming_level": info["programming_level"],
        "mathematics_level": info["mathematics_level"],
        "communication_level": info["communication_level"],
        "problem_solving_level": info["problem_solving_level"],
        "average_salary": (
            round(salary_info["average_salary"], 2)
            if salary_info else "Not available"
        ),
        "salary_range": (
            f"₹{salary_info['minimum_salary']:,.0f} - "
            f"₹{salary_info['maximum_salary']:,.0f}"
            if salary_info else "Not available"
        ),
        "roadmap": roadmaps.get(
            career,
            ["Build relevant skills", "Create projects", "Apply for jobs"]
        )
    }


# ------------------------------------------------------------
# 7. PREPARE RECOMMENDATION DATA
# ------------------------------------------------------------

# If matching_skills are not already stored in the recommendation
# file, calculate them from the career knowledge base.

if "matching_skills" not in recommendations.columns:

    matching_values = []

    # Example skills used in our test profile
    student_skills = [
        "python",
        "sql",
        "machine learning",
        "research"
    ]

    for career in recommendations["career"]:

        info = get_career_info(career)

        if info is None:
            matching_values.append("")
            continue

        career_skills = [
            skill.strip().lower()
            for skill in str(info["skills"]).split(",")
        ]

        matches = [
            skill
            for skill in student_skills
            if skill.lower() in career_skills
        ]

        matching_values.append(", ".join(matches))

    recommendations["matching_skills"] = matching_values


# ------------------------------------------------------------
# 8. GENERATE TOP CAREER EXPLANATION
# ------------------------------------------------------------

top_row = recommendations.iloc[0]

result = generate_explanation(top_row)


# ------------------------------------------------------------
# 9. DISPLAY FINAL RESULT
# ------------------------------------------------------------

print("\n")
print("=" * 80)
print("PERSONALIZED CAREER RECOMMENDATION")
print("=" * 80)

print(f"\nRecommended Career : {result['career']}")
print(f"Match Score        : {result['score']}%")
print(f"Career Category    : {result['category']}")

print("\nWhy this career?")
print("-" * 40)
print(result["explanation"])

print("\nYour Matching Skills")
print("-" * 40)
print(result["matching_skills"])

print("\nSkills to Improve")
print("-" * 40)
print(result["skills_to_improve"])

print("\nEducation Requirement")
print("-" * 40)
print(result["education"])

print("\nExpected Salary Information")
print("-" * 40)
print(f"Average Salary : ₹{result['average_salary']:,.2f}"
      if isinstance(result["average_salary"], (int, float))
      else f"Average Salary : {result['average_salary']}")
print(f"Salary Range   : {result['salary_range']}")

print("\n")
print("=" * 80)
print("CAREER ROADMAP")
print("=" * 80)

for i, step in enumerate(result["roadmap"], start=1):
    print(f"{i}. {step}")

print("\n")
print("=" * 80)
print("STEP 9 COMPLETED SUCCESSFULLY")
print("=" * 80)
```
