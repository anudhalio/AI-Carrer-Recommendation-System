```python
import pandas as pd
import os
import glob

# ============================================================
# AI CAREER RECOMMENDATION SYSTEM
# DATASET ANALYSIS
# ============================================================

print("=" * 70)
print("AI CAREER RECOMMENDATION SYSTEM - DATASET ANALYSIS")
print("=" * 70)

# ------------------------------------------------------------
# 1. LOCATE DATASET
# ------------------------------------------------------------

dataset_folder = r"data"

print("\nSearching inside:")
print(os.path.abspath(dataset_folder))

# Search recursively for CSV/Excel files
files = glob.glob(os.path.join(dataset_folder, "**", "*.csv"), recursive=True)
files += glob.glob(os.path.join(dataset_folder, "**", "*.xlsx"), recursive=True)
files += glob.glob(os.path.join(dataset_folder, "**", "*.xls"), recursive=True)

print("\nFiles found:")

if len(files) == 0:
    print("❌ No CSV or Excel file found.")
    print("\nPlease check what is inside your data folder.")
else:
    for i, file in enumerate(files, 1):
        print(f"{i}. {os.path.abspath(file)}")

# ------------------------------------------------------------
# 2. LOAD DATASET
# ------------------------------------------------------------

if len(files) == 0:
    raise FileNotFoundError(
        "\nDataset file could not be found."
    )

# Use the first dataset file
dataset_path = files[0]

print("\n" + "=" * 70)
print("LOADING DATASET")
print("=" * 70)

print("Selected file:")
print(dataset_path)

if dataset_path.lower().endswith(".csv"):
    df = pd.read_csv(dataset_path)
else:
    df = pd.read_excel(dataset_path)

print("\n✅ Dataset loaded successfully!")

# ------------------------------------------------------------
# 3. BASIC INFORMATION
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("1. BASIC DATASET INFORMATION")
print("=" * 70)

print(f"Rows    : {df.shape[0]:,}")
print(f"Columns : {df.shape[1]}")

print("\nColumn names:")
for i, col in enumerate(df.columns, 1):
    print(f"{i}. {col}")

# ------------------------------------------------------------
# 4. FIRST 10 ROWS
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("2. FIRST 10 ROWS")
print("=" * 70)

display(df.head(10))

# ------------------------------------------------------------
# 5. DATA TYPES
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("3. DATA TYPES")
print("=" * 70)

display(df.dtypes.to_frame("Data Type"))

# ------------------------------------------------------------
# 6. MISSING VALUES
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("4. MISSING VALUES")
print("=" * 70)

missing = df.isnull().sum()

missing_table = pd.DataFrame({
    "Column": df.columns,
    "Missing Values": missing.values,
    "Missing %": (missing.values / len(df) * 100).round(2)
})

display(missing_table)

# ------------------------------------------------------------
# 7. DUPLICATES
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("5. DUPLICATE ROWS")
print("=" * 70)

duplicate_count = df.duplicated().sum()

print(f"Duplicate rows: {duplicate_count:,}")

# ------------------------------------------------------------
# 8. UNIQUE VALUES
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("6. UNIQUE VALUES PER COLUMN")
print("=" * 70)

unique_table = pd.DataFrame({
    "Column": df.columns,
    "Unique Values": [
        df[col].nunique(dropna=True)
        for col in df.columns
    ]
})

display(unique_table)

# ------------------------------------------------------------
# 9. JOB TITLE ANALYSIS
# ------------------------------------------------------------

if "Job Title" in df.columns:

    print("\n" + "=" * 70)
    print("7. JOB TITLE ANALYSIS")
    print("=" * 70)

    print(
        "Unique Job Titles:",
        df["Job Title"].nunique()
    )

    print("\nTop 20 Job Titles:")

    display(
        df["Job Title"]
        .value_counts()
        .head(20)
        .to_frame("Number of Jobs")
    )

# ------------------------------------------------------------
# 10. COMPANY ANALYSIS
# ------------------------------------------------------------

if "Company" in df.columns:

    print("\n" + "=" * 70)
    print("8. COMPANY ANALYSIS")
    print("=" * 70)

    print(
        "Unique Companies:",
        df["Company"].nunique()
    )

# ------------------------------------------------------------
# 11. LOCATION ANALYSIS
# ------------------------------------------------------------

if "Location" in df.columns:

    print("\n" + "=" * 70)
    print("9. LOCATION ANALYSIS")
    print("=" * 70)

    print(
        "Unique Locations:",
        df["Location"].nunique()
    )

    display(
        df["Location"]
        .value_counts()
        .head(20)
        .to_frame("Number of Jobs")
    )

# ------------------------------------------------------------
# 12. EXPERIENCE LEVEL
# ------------------------------------------------------------

if "Experience Level" in df.columns:

    print("\n" + "=" * 70)
    print("10. EXPERIENCE LEVEL")
    print("=" * 70)

    display(
        df["Experience Level"]
        .value_counts()
        .to_frame("Number of Jobs")
    )

# ------------------------------------------------------------
# 13. INDUSTRY ANALYSIS
# ------------------------------------------------------------

if "Industry" in df.columns:

    print("\n" + "=" * 70)
    print("11. INDUSTRY ANALYSIS")
    print("=" * 70)

    print(
        "Unique Industries:",
        df["Industry"].nunique()
    )

    display(
        df["Industry"]
        .value_counts()
        .to_frame("Number of Jobs")
    )

# ------------------------------------------------------------
# 14. SALARY ANALYSIS
# ------------------------------------------------------------

if "Salary" in df.columns:

    print("\n" + "=" * 70)
    print("12. SALARY ANALYSIS")
    print("=" * 70)

    salary = pd.to_numeric(
        df["Salary"],
        errors="coerce"
    )

    print(f"Minimum Salary : {salary.min():,.0f}")
    print(f"Maximum Salary : {salary.max():,.0f}")
    print(f"Average Salary : {salary.mean():,.2f}")
    print(f"Median Salary  : {salary.median():,.2f}")

# ------------------------------------------------------------
# 15. REQUIRED SKILLS
# ------------------------------------------------------------

if "Required Skills" in df.columns:

    print("\n" + "=" * 70)
    print("13. REQUIRED SKILLS ANALYSIS")
    print("=" * 70)

    skills = (
        df["Required Skills"]
        .dropna()
        .astype(str)
        .str.split(",")
        .explode()
        .str.strip()
    )

    skills = skills[skills != ""]

    print(
        "Total Skill Occurrences:",
        len(skills)
    )

    print(
        "Unique Skills:",
        skills.nunique()
    )

    print("\nTop 30 Skills:")

    display(
        skills
        .value_counts()
        .head(30)
        .to_frame("Number of Jobs")
    )

# ------------------------------------------------------------
# 16. JOB + SKILLS EXAMPLES
# ------------------------------------------------------------

if "Job Title" in df.columns and "Required Skills" in df.columns:

    print("\n" + "=" * 70)
    print("14. JOB TITLE → REQUIRED SKILLS")
    print("=" * 70)

    display(
        df[["Job Title", "Required Skills"]]
        .head(20)
    )

# ------------------------------------------------------------
# 17. JOB + INDUSTRY EXAMPLES
# ------------------------------------------------------------

if "Job Title" in df.columns and "Industry" in df.columns:

    print("\n" + "=" * 70)
    print("15. JOB TITLE → INDUSTRY")
    print("=" * 70)

    display(
        df[["Job Title", "Industry"]]
        .head(20)
    )

# ------------------------------------------------------------
# 18. FINAL DATA QUALITY SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("16. FINAL DATA QUALITY SUMMARY")
print("=" * 70)

print(f"Rows                  : {len(df):,}")
print(f"Columns               : {len(df.columns)}")
print(f"Duplicate Rows        : {df.duplicated().sum():,}")
print(f"Total Missing Values  : {df.isnull().sum().sum():,}")

if "Job Title" in df.columns:
    print(
        f"Unique Job Titles     : "
        f"{df['Job Title'].nunique():,}"
    )

if "Industry" in df.columns:
    print(
        f"Unique Industries     : "
        f"{df['Industry'].nunique():,}"
    )

if "Location" in df.columns:
    print(
        f"Unique Locations      : "
        f"{df['Location'].nunique():,}"
    )

if "Company" in df.columns:
    print(
        f"Unique Companies      : "
        f"{df['Company'].nunique():,}"
    )

if "Required Skills" in df.columns:
    print(
        f"Unique Skills         : "
        f"{skills.nunique():,}"
    )

print("\n" + "=" * 70)
print("✅ DATASET ANALYSIS COMPLETED")
print("=" * 70)
```

```python
# ============================================================
# STEP 3: CLEAN AND PREPARE JOB DATASET
# ============================================================

import pandas as pd
import numpy as np
import re
import os

print("=" * 70)
print("STEP 3 - DATA CLEANING AND PREPARATION")
print("=" * 70)

# ------------------------------------------------------------
# 1. LOAD ORIGINAL DATASET
# ------------------------------------------------------------

input_path = r"data/job_recommendation_dataset.csv"

df = pd.read_csv(input_path)

print("\nOriginal dataset:")
print(f"Rows    : {len(df):,}")
print(f"Columns : {len(df.columns)}")

# ------------------------------------------------------------
# 2. CREATE COPY
# ------------------------------------------------------------

jobs = df.copy()

# ------------------------------------------------------------
# 3. STANDARDIZE COLUMN NAMES
# ------------------------------------------------------------

jobs.columns = [
    col.strip().lower().replace(" ", "_")
    for col in jobs.columns
]

print("\nStandardized columns:")
print(list(jobs.columns))

# ------------------------------------------------------------
# 4. CLEAN TEXT COLUMNS
# ------------------------------------------------------------

text_columns = [
    "job_title",
    "company",
    "location",
    "experience_level",
    "industry",
    "required_skills"
]

for col in text_columns:
    if col in jobs.columns:
        jobs[col] = (
            jobs[col]
            .astype(str)
            .str.strip()
            .str.replace(r"\s+", " ", regex=True)
        )

# ------------------------------------------------------------
# 5. CLEAN SALARY
# ------------------------------------------------------------

jobs["salary"] = pd.to_numeric(
    jobs["salary"],
    errors="coerce"
)

# ------------------------------------------------------------
# 6. CLEAN REQUIRED SKILLS
# ------------------------------------------------------------

def clean_skills(skill_text):
    if pd.isna(skill_text):
        return ""
    
    skills = str(skill_text).split(",")
    
    cleaned = []
    
    for skill in skills:
        skill = skill.strip()
        skill = re.sub(r"\s+", " ", skill)
        
        if skill:
            cleaned.append(skill)
    
    # Remove duplicate skills within the same row
    cleaned = list(dict.fromkeys(cleaned))
    
    return ", ".join(cleaned)


jobs["required_skills"] = jobs["required_skills"].apply(
    clean_skills
)

# ------------------------------------------------------------
# 7. REMOVE EXACT DUPLICATES
# ------------------------------------------------------------

before = len(jobs)

jobs = jobs.drop_duplicates()

after = len(jobs)

print("\nDuplicate removal:")
print(f"Rows before : {before:,}")
print(f"Rows after  : {after:,}")
print(f"Removed     : {before - after:,}")

# ------------------------------------------------------------
# 8. REMOVE INVALID ROWS
# ------------------------------------------------------------

required_columns = [
    "job_title",
    "location",
    "experience_level",
    "salary",
    "industry",
    "required_skills"
]

before = len(jobs)

jobs = jobs.dropna(
    subset=required_columns
)

jobs = jobs[
    (jobs["job_title"].str.len() > 1) &
    (jobs["required_skills"].str.len() > 0)
]

after = len(jobs)

print("\nInvalid row removal:")
print(f"Rows before : {before:,}")
print(f"Rows after  : {after:,}")
print(f"Removed     : {before - after:,}")

# ------------------------------------------------------------
# 9. RESET INDEX
# ------------------------------------------------------------

jobs = jobs.reset_index(drop=True)

# ------------------------------------------------------------
# 10. CREATE JOB ID
# ------------------------------------------------------------

jobs.insert(
    0,
    "job_id",
    range(1, len(jobs) + 1)
)

# ------------------------------------------------------------
# 11. CREATE SKILL LIST COLUMN
# ------------------------------------------------------------

jobs["skill_list"] = jobs["required_skills"].apply(
    lambda x: [
        skill.strip()
        for skill in x.split(",")
        if skill.strip()
    ]
)

# ------------------------------------------------------------
# 12. CREATE SKILL COUNT
# ------------------------------------------------------------

jobs["skill_count"] = jobs["skill_list"].apply(len)

# ------------------------------------------------------------
# 13. CREATE SALARY CATEGORY
# ------------------------------------------------------------

def salary_category(salary):
    
    if salary < 70000:
        return "Low"
    
    elif salary < 100000:
        return "Medium"
    
    elif salary < 125000:
        return "High"
    
    else:
        return "Very High"


jobs["salary_category"] = jobs["salary"].apply(
    salary_category
)

# ------------------------------------------------------------
# 14. CREATE EXPERIENCE SCORE
# ------------------------------------------------------------

experience_score = {
    "Entry Level": 1,
    "Mid Level": 2,
    "Senior Level": 3
}

jobs["experience_score"] = jobs[
    "experience_level"
].map(experience_score)

# ------------------------------------------------------------
# 15. CREATE SEARCHABLE JOB PROFILE
# ------------------------------------------------------------

jobs["job_profile"] = (
    jobs["job_title"].fillna("") + " " +
    jobs["industry"].fillna("") + " " +
    jobs["required_skills"].fillna("")
)

# ------------------------------------------------------------
# 16. SAVE CLEAN DATASET
# ------------------------------------------------------------

output_path = r"data/processed_jobs.csv"

jobs.to_csv(
    output_path,
    index=False
)

print("\n" + "=" * 70)
print("CLEAN DATASET CREATED")
print("=" * 70)

print(f"\nSaved to:")
print(os.path.abspath(output_path))

print(f"\nFinal rows    : {len(jobs):,}")
print(f"Final columns : {len(jobs.columns)}")

# ------------------------------------------------------------
# 17. SHOW FINAL COLUMNS
# ------------------------------------------------------------

print("\nFinal columns:")

for i, col in enumerate(jobs.columns, 1):
    print(f"{i}. {col}")

# ------------------------------------------------------------
# 18. SHOW SAMPLE
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("SAMPLE OF CLEAN DATA")
print("=" * 70)

display(
    jobs[
        [
            "job_id",
            "job_title",
            "location",
            "experience_level",
            "salary",
            "industry",
            "required_skills",
            "skill_count",
            "salary_category"
        ]
    ].head(10)
)

# ------------------------------------------------------------
# 19. FINAL CHECK
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("FINAL CHECK")
print("=" * 70)

print(
    "Missing values:",
    jobs.isnull().sum().sum()
)

print(
    "Duplicate rows:",
    jobs.duplicated().sum()
)

print(
    "Unique careers:",
    jobs["job_title"].nunique()
)

print(
    "Unique industries:",
    jobs["industry"].nunique()
)

print(
    "Unique locations:",
    jobs["location"].nunique()
)

print(
    "Average salary:",
    f"{jobs['salary'].mean():,.2f}"
)

print("\n✅ STEP 3 COMPLETED SUCCESSFULLY!")
```
